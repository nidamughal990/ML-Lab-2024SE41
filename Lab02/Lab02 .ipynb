### (c) Find students who passed only because of grace marks

```python
passed_only_after_grace = (marks < 50) & (grace_adjusted_marks >= 50)

student_indices = np.where(passed_only_after_grace)[0]

print("T9c: Indices of students who passed only because of grace marks:",
      student_indices)
```

### (d) Calculate the z-score of each student's marks after grace

```python
mean_marks = np.mean(grace_adjusted_marks)
std_marks = np.std(grace_adjusted_marks)

z_scores = (grace_adjusted_marks - mean_marks) / std_marks

print("T9d: Mean marks:", mean_marks)
print("T9d: Standard deviation:", std_marks)
print("T9d: Z-scores:", z_scores)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

### (a) Create an 8 × 8 image and draw the number 7


```python
img = np.zeros((8, 8), dtype=int)

# Top horizontal line
img[1, 1:7] = 255

# Straight vertical line on the right
img[1:7, 5] = 255

# Middle horizontal crossbar
img[4, 3:7] = 255

print("Original Image:")
print(img)
## display the number

plt.figure(figsize=(6, 6))
plt.imshow(img, cmap='gray', vmin=0, vmax=255,
           interpolation='nearest')
plt.title("Original Image - Number 7")
plt.axis('off')
plt.show()
```

### (b) Flip the image from left to right

```python
flipped_img = np.fliplr(img)

print("T10b: Flipped Image:")
print(flipped_img)

# Display flipped image
plt.figure(figsize=(5, 5))
plt.imshow(flipped_img, cmap='gray', vmin=0, vmax=255,
           interpolation='nearest')
plt.title("Flipped Image")
plt.axis('off')
plt.show()
```

### (c) Invert the image

```python
inverted_img = 255 - img

print("T10c: Inverted Image:")
print(inverted_img)

# Display inverted image
plt.figure(figsize=(5, 5))
plt.imshow(inverted_img, cmap='gray', vmin=0, vmax=255,
           interpolation='nearest')
plt.title("Inverted Image")
plt.axis('off')
plt.show()
```

### (d) Crop rows 1–4 and columns 3–6
Python uses zero-based indexing, and the ending index is excluded.

```python
cropped_img = img[1:5, 3:7]

print("T10d: Cropped Image:")
print(cropped_img)

# Display cropped image
plt.figure(figsize=(5, 5))
plt.imshow(cropped_img, cmap='gray', vmin=0, vmax=255,
           interpolation='nearest')
plt.title("Cropped Image")
plt.axis('off')
plt.show()
```

### (e) Brighten the image by 80 and cap values at 255

```python
bright_img = np.clip(img + 80, 0, 255)

print("T10e: Brightened Image:")
print(bright_img)

# Display brightened image
plt.figure(figsize=(5, 5))
plt.imshow(bright_img, cmap='gray', vmin=0, vmax=255,
           interpolation='nearest')
plt.title("Brightened Image")
plt.axis('off')
plt.show()
```

### (f) Print the image's shape, data type, and number of white pixels

```python
print("T10f: Image Properties")

# Print image shape
print("Image Shape:", img.shape)

# Print image data type
print("Image Data Type:", img.dtype)

# Count white pixels (value = 255)
print("Number of White Pixels:", np.sum(img == 255))
```
